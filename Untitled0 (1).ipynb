````python

df['Application Link'] = df['Application Link'].fillna('Not available')

print(df.isnull().sum())
````

````python

# Convertir chaque scholarship en un texte (Chunk)

chunks = []

for index, row in df.iterrows():
    chunk = f"""
    Name: {row['Name']}
    Sponsor: {row['Sponsor']}
    Type: {row['Type']}
    Region: {row['Region']}
    Study Level: {row['Study Level']}
    Amount: {row['Amount']}
    Deadline: {row['Deadline']}
    Application Link: {row['Application Link']}
    """

    chunks.append(chunk)

print("Nombre de chunks :", len(chunks))
print("\nPremier chunk :")
print(chunks[0])
````

````python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2')
````

````python
embeddings = model.encode(chunks, show_progress_bar=True)

print("Shape des embeddings :", embeddings.shape)
````

````python
!pip install faiss-cpu
````

````python
import faiss
import numpy as np

embeddings = np.array(embeddings).astype('float32')

dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(embeddings)

print("Nombre de vectors dans FAISS :", index.ntotal)
````

````python

def retrieve_context(question, k=3):
    question_embedding = model.encode([question])
    question_embedding = np.array(question_embedding).astype('float32')

    distances, indices = index.search(question_embedding, k)

    results = []
    for idx in indices[0]:
        results.append(chunks[idx])

    return results

print("Retrieval function is ready!")
````

````python
for rank, idx in enumerate(indices[0]):
    print(f"Resultat {rank + 1}")
    print("Distance :", distances[0][rank])
    print(chunks[idx])
    print("-" * 50)
````

````python
def generate_answer(question, context):

    prompt = f"""
    You are a scholarship assistant.

    Answer the user's question using ONLY the information
    provided in the context below.

    If the answer is not available in the context,
    say: "I don't have enough information in my database."

    Always mention the scholarship names and application links
    when available.

    Context:
    {context}

    User Question:
    {question}

    Answer:
    """

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return response.text
````

````python
def rag_chatbot(question):

    # Step 1: Retrieve relevant information
    context = retrieve_context(question)

    # Step 2: Generate answer using Gemini
    answer = generate_answer(question, "\n".join(context))

    return answer
````

### Extract Sources

````python
def extract_sources(context):

    sources = []

    for chunk in context:
        name = "Unknown"
        link = "Not available"

        for line in chunk.splitlines():
            line = line.strip()

            if line.startswith("Name:"):
                name = line.replace("Name:", "").strip()

            elif line.startswith("Application Link:"):
                link = line.replace("Application Link:", "").strip()

        source = {
            "name": name,
            "link": link
        }

        if source not in sources:
            sources.append(source)

    return sources
````

### RAG Chatbot with Sources

````python
def rag_chatbot_with_sources(question):

    # 1. Retrieve relevant chunks
    context = retrieve_context(question)

    # 2. Generate answer
    answer = generate_answer(question, "\n".join(context))

    # 3. Extract all sources
    all_sources = extract_sources(context)

    # 4. Keep only sources mentioned in the answer
    if "I don't have enough information" in answer:
        sources = []
    else:
        sources = [
            source for source in all_sources
            if source["name"].lower() in answer.lower()
        ]

    # 5. Return answer and sources
    return answer, sources
````

````python
def chatbot_interface(question):
    answer, sources = rag_chatbot_with_sources(question)

    sources_text = "\n".join([
        f"- {source['name']}: {source['link']}"
        for source in sources
    ])

    return answer, sources_text
````

````python
import gradio as gr

demo = gr.Interface(
    fn=chatbot_interface,
    inputs=gr.Textbox(
        label="Ask your scholarship question",
        placeholder="Example: What scholarships are available for undergraduate students?"
    ),
    outputs=[
        gr.Textbox(label="Chatbot Answer"),
        gr.Textbox(label="Sources")
    ],
    title="Scholarship Finder Chatbot",
    description="Ask questions about scholarships using our collected dataset."
)

demo.launch(share=True)
````
